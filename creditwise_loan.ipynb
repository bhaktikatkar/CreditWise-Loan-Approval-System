```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv("loan_approval_data.csv")
```

```python
df.head()
df.info()
df.isnull().sum 
df.describe() # to observe numerical value 
```

```python
#handel missing value
categorial_cols = df.select_dtypes(include=["object"]).columns
numerical_cols = df.select_dtypes(include=["float64"]).columns
```

```python
categorial_cols.size + numerical_cols.size
```

```python
from sklearn.impute import SimpleImputer

num_imp = SimpleImputer(strategy="mean")
df[numerical_cols]= num_imp.fit_transform(df[numerical_cols])
```

```python
cat_imp = SimpleImputer(strategy="most_frequent")
df[categorial_cols]= cat_imp.fit_transform(df[categorial_cols])
```

```python
df.head()
df.isnull().sum()
```

```python
#EDA-TO ANALYSE DIFF FEATURES
classes_count=df["Loan_Approved"].value_counts()
plt.pie(classes_count, labels=["no","yes"],autopct="%1.1f%%")
plt.title("is loan approved or not?")       
```

```python
# to analyze categories 

gender_cnt= df["Gender"].value_counts()
ax=sns.barplot(gender_cnt)
ax.bar_label(ax.containers[0])
```

```python
edu_cnt= df["Education_Level"].value_counts()
ax=sns.barplot(edu_cnt)
ax.bar_label(ax.containers[0])
```
